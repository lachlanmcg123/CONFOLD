```python
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score
from collections import Counter

# Import the custom modules from your project
from datasets import annaburroo
from utils import split_xy
from algo import prune_rules

# Load the Annaburroo dataset and the untrained Classifier model
# The annaburroo() function handles the file reading and Classifier instantiation
confold_model, data = annaburroo()

print("Setup complete. Annaburroo dataset is loaded.")
```

```python
### Demonstration of Early Termination
# 1. Split the data manually: First 100 for training, the remainder for testing
# (Assuming 'data' is returned as a list of records by the Classifier.load_data method)
data_train = data[:100]
data_test = data[100:]

print("Training Set Class Counts:", dict(Counter(row[-1] for row in data_train)))
print("Testing Set Class Counts: ", dict(Counter(row[-1] for row in data_test)))

# Extract features (X) and targets (Y) for the test set
X_test, Y_test = split_xy(data_test)

# 2. Train the CON-FOLD model on the training data, no pruning used at all
confold_model.confidence_fit(data_train, improvement_threshold=-1) #Setting threshold to -1 means that any exception will be kept no matter how little it improves

if hasattr(confold_model, 'asp'):
    confold_model.asp()

# 3. Print the generated rules with readable formatting
print("--- Generated CON-FOLD Rules ---")
if hasattr(confold_model, 'crs') and confold_model.crs:
    features = confold_model.attrs 
    for i, rule in enumerate(confold_model.crs):
        if rule[-1] > 0:
            formatted_conditions = []
            for cond in rule[1]:
                feat_idx, operator, value = cond
                if isinstance(feat_idx, int):
                    feat_name = features[feat_idx]
                else:
                    feat_name = feat_idx
                formatted_conditions.append(f"{feat_name} {operator} {value}")
            body_str = " AND ".join(formatted_conditions)
            class_name = rule[0][2]
            head_str = f"Class = '{class_name}'"
            print(f"Rule {i+1}: IF {body_str} THEN {head_str} (Confidence: {rule[-1]:.4f})")
else:
    print("No rules were generated.")

# 4. Make predictions and evaluate accuracy on the test set
predictions_with_confidence = confold_model.predict(X_test)

# Extract the predicted class, using 'Unknown' as a fallback if no rule matched
predictions = [p[0] if p[0] is not None else 'Unknown' for p in predictions_with_confidence]

accuracy = accuracy_score(Y_test, predictions)
print("\n--- Model Performance ---")
print(f"Accuracy on the test set: {accuracy:.4f}")

# 4. Make predictions and evaluate accuracy on the test set
X_test, Y_test = split_xy(data_train)
predictions_with_confidence = confold_model.predict(X_test)

# Extract the predicted class, using 'Unknown' as a fallback if no rule matched
predictions = [p[0] if p[0] is not None else 'Unknown' for p in predictions_with_confidence]

accuracy = accuracy_score(Y_test, predictions)
print("\n--- Model Performance on training data ---")
print(f"Accuracy on the test set: {accuracy:.4f}")
```

```python
#Additional imports for experiments
#!pip install xgboost

import numpy as np
import pandas as pd
import time
import sqlite3
from joblib import Parallel, delayed
from IPython.display import display

import xgboost as xgb
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score

# Import your project files
from foldrm import Classifier
from utils import split_data, split_xy, get_scores, count_rules_in_model
from datasets import annaburroo
from algo import prune_rules

# --- Experiment Configuration ---
datasets_with_names = [(annaburroo, "Annaburroo")]

# Define the fit methods and strategies to test
fit_methods = ["FOLD-RM (fit)", "CON-FOLD (confidence_fit)"]
selection_strategies = ['greedy', 'round_robin', 'best_rule', 'info_gain']

metric_list = ['original', 'original_guarded', 'original_reweighted', 'original_t_removed', 'IG_over_P', 'IG_over_P2', 'LogOddsRatio',  'MathewsCC',
               'MathewsCC_mod', 'MathewsP', 'Gini_Impurity', 'Precision_Gini_Impurity']

NUM_TRIALS = 300 
DB_FILE = 'annaburroo_concise_exp.db'

# Display options for the final DataFrame
pd.set_option('display.max_rows', 500)
pd.set_option('display.max_columns', 50)
pd.set_option('display.width', 1000)
```

```python
#Helper Functions for Experiments
def run_one_trial(dataset_name, strategy, fit_method, metric, data, model_template, num_classes, db_file):
    start_time = time.time()
    
    # 50/50 Train/Test split
    data_train, data_test = split_data(data, ratio=0.5) 
    X_train, Y_train = split_xy(data_train)
    X_test, Y_test = split_xy(data_test)
    
    rule_count = 0
    score = 0
    
    if fit_method == "XGBoost":
        # Ensure X features are purely numeric (stripping out string IDs like FireID)
        X_train_num = [[float(x) for x in row if isinstance(x, (int, float)) or (isinstance(x, str) and x.replace('.','',1).replace('-','',1).isdigit())] for row in X_train]
        X_test_num = [[float(x) for x in row if isinstance(x, (int, float)) or (isinstance(x, str) and x.replace('.','',1).replace('-','',1).isdigit())] for row in X_test]
        
        # Encode string labels to integers for XGBoost
        le = LabelEncoder()
        # Extract target strings whether they are tuples or bare strings
        y_train_flat = [y[0] if isinstance(y, tuple) else y for y in Y_train]
        y_test_flat = [y[0] if isinstance(y, tuple) else y for y in Y_test]
        
        le.fit(y_train_flat + y_test_flat)
        y_tr_enc = le.transform(y_train_flat)
        y_te_enc = le.transform(y_test_flat)
        
        # Train and Predict with default parameters
        clf = xgb.XGBClassifier(eval_metric='mlogloss')
        clf.fit(np.array(X_train_num), y_tr_enc)
        y_pred_enc = clf.predict(np.array(X_test_num))
        
        score = accuracy_score(y_te_enc, y_pred_enc)
        rule_count = 0 # XGBoost uses trees, so we default to 0 for the rule output
        
    else:
        # FOLD-RM and CON-FOLD Logic
        model = Classifier(attrs=model_template.attrs, numeric=model_template.numeric, label=model_template.label)
        
        if fit_method == "CON-FOLD (confidence_fit)":
            model.confidence_fit(data_train, metric=metric, num_classes=num_classes, selection_strategy=strategy)
        else:
            model.fit(data_train, metric=metric, num_classes=num_classes, selection_strategy=strategy)
        
        Ystar_test_tuples = model.predict(X_test)
        Ystar_test = [y[0] if y[0] is not None else 'Unknown' for y in Ystar_test_tuples]
        
        score = get_scores(Ystar_test, data_test)
        rule_count = count_rules_in_model(model)
        
    elapsed_time = time.time() - start_time
    
    # Save to database
    conn = sqlite3.connect(db_file)
    cursor = conn.cursor()
    cursor.execute('PRAGMA journal_mode=WAL;')
    cursor.execute('''
        INSERT INTO trials (dataset, strategy, fit_method, metric, accuracy, time, num_rules)
        VALUES (?, ?, ?, ?, ?, ?, ?)
    ''', (dataset_name, strategy, fit_method, metric, score, elapsed_time, rule_count))
    conn.commit()
    conn.close()

def run_experiments_for_combination(dataset_info, strategy, fit_method, metric, remaining_trials, db_file):
    dataset_func, name = dataset_info
    print(f"--- Starting: Method='{fit_method}', Metric='{metric}', Strategy='{strategy}', Remaining={remaining_trials} ---")
    
    model_template, data = dataset_func()
    num_classes = len(pd.unique(pd.DataFrame(data).iloc[:, -1]))
    
    if remaining_trials > 0:
        Parallel(n_jobs=-1, verbose=0)(
            delayed(run_one_trial)(name, strategy, fit_method, metric, data, model_template, num_classes, db_file)
            for _ in range(remaining_trials)
        )
```

```python
##Main experiment run
if __name__ == "__main__":
    # Create database and table
    conn = sqlite3.connect(DB_FILE)
    cursor = conn.cursor()
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS trials (
            dataset TEXT,
            strategy TEXT,
            fit_method TEXT,
            metric TEXT,
            accuracy REAL,
            time REAL,
            num_rules INTEGER
        )
    ''')
    conn.commit()

    # Get current counts
    counts = {}
    cursor.execute('SELECT fit_method, metric, strategy, COUNT(*) FROM trials GROUP BY fit_method, metric, strategy')
    for row in cursor.fetchall():
        fm, met, strat, cnt = row
        counts[(fm, met, strat)] = cnt
    conn.close()

    tasks_to_run = []
    
    # 1. Add FOLD tasks
    for dataset_info in datasets_with_names:
        for fm in fit_methods:
            for metric in metric_list:
                for strategy in selection_strategies:
                    key = (fm, metric, strategy)
                    remaining = NUM_TRIALS - counts.get(key, 0)
                    if remaining > 0:
                        tasks_to_run.append({
                            'dataset_info': dataset_info, 'strategy': strategy, 
                            'fit_method': fm, 'metric': metric, 
                            'remaining_trials': remaining, 'db_file': DB_FILE
                        })
                        
    # 2. Add XGBoost Control task
    xgb_key = ("XGBoost", "XGBoost", "Control")
    xgb_remaining = NUM_TRIALS - counts.get(xgb_key, 0)
    if xgb_remaining > 0:
        tasks_to_run.append({
            'dataset_info': datasets_with_names[0], 'strategy': "Control", 
            'fit_method': "XGBoost", 'metric': "XGBoost", 
            'remaining_trials': xgb_remaining, 'db_file': DB_FILE
        })

    print(f"\n>>> Found {len(tasks_to_run)} combinations with remaining trials. <<<\n")

    if tasks_to_run:
        for task in tasks_to_run:
            run_experiments_for_combination(**task)
        print("\n--- All tasks complete! You can now run the aggregation cell. ---")
    else:
        print("\n--- No remaining trials to run. Everything is complete. ---")
```

```python
#Display Results
def format_with_ste(mean_val, std_val, count):
    """Formats values as $mean \pm STE$."""
    ste = std_val / np.sqrt(count)
    return f"${mean_val:.4f} \pm {ste:.4f}$"

def process_and_display_concise(df, xgb_df, fit_method_name):
    if df.empty:
        return

    # Aggregate Data
    agg = df.groupby(['metric', 'strategy']).agg({
        'accuracy': ['mean', 'std', 'count'],
        'num_rules': ['mean', 'std', 'count'],
        'time': ['mean', 'std', 'count']
    })

    # Format values with STE
    for col in ['accuracy', 'num_rules', 'time']:
        agg[(col, 'formatted')] = agg.apply(lambda row: format_with_ste(row[(col, 'mean')], row[(col, 'std')], row[(col, 'count')]), axis=1)

    # Pivot Tables
    acc_pivot = agg['accuracy']['formatted'].unstack(level='strategy')
    rules_pivot = agg['num_rules']['formatted'].unstack(level='strategy')
    time_pivot = agg['time']['formatted'].unstack(level='strategy')
    
    # Order the columns safely
    ordered_cols = [s for s in selection_strategies if s in acc_pivot.columns]
    acc_pivot = acc_pivot[ordered_cols]
    rules_pivot = rules_pivot[ordered_cols]
    time_pivot = time_pivot[ordered_cols]

    # Combine into wide format
    final_table = pd.concat([acc_pivot, rules_pivot, time_pivot], axis=1, keys=['Accuracy', 'Number of Rules', 'Time'])

    # Process and append XGBoost Control Row
    if not xgb_df.empty:
        xgb_agg = xgb_df.agg({
            'accuracy': ['mean', 'std', 'count'],
            'num_rules': ['mean', 'std', 'count'],
            'time': ['mean', 'std', 'count']
        })
        
        xgb_row = {}
        for top_col in ['Accuracy', 'Number of Rules', 'Time']:
            db_col = top_col.lower().replace('number of rules', 'num_rules')
            
            # Use standard formatting for rules vs accuracy/time
            if db_col == 'num_rules':
                val_str = "-" # XGBoost has no rules
            else:
                val_str = format_with_ste(xgb_agg.loc['mean', db_col], xgb_agg.loc['std', db_col], xgb_agg.loc['count', db_col])
                
            for strat in ordered_cols:
                xgb_row[(top_col, strat)] = val_str # Duplicate across row for easy visual comparison
                
        xgb_series = pd.Series(xgb_row, name="XGBoost (Control)")
        final_table = pd.concat([final_table, xgb_series.to_frame().T])

    print(f"\n{'='*60}")
    print(f"--- STRATEGY EXPERIMENT SUMMARY: {fit_method_name} ---")
    
    # Display table (Jupyter will automatically render the LaTeX \pm signs)
    display(final_table)
    print(f"{'='*60}")

if __name__ == "__main__":
    conn = sqlite3.connect(DB_FILE)
    try:
        all_trials_df = pd.read_sql_query("SELECT * FROM trials", conn)
    except pd.io.sql.DatabaseError:
        all_trials_df = pd.DataFrame()
    conn.close()

    if not all_trials_df.empty:
        # Separate out the XGBoost data
        xgb_df = all_trials_df[all_trials_df['fit_method'] == 'XGBoost']
        
        # Process FOLD-RM
        fold_rm_df = all_trials_df[all_trials_df['fit_method'] == 'FOLD-RM (fit)']
        process_and_display_concise(fold_rm_df, xgb_df, "FOLD-RM (fit)")
        
        # Process CON-FOLD
        con_fold_df = all_trials_df[all_trials_df['fit_method'] == 'CON-FOLD (confidence_fit)']
        process_and_display_concise(con_fold_df, xgb_df, "CON-FOLD (confidence_fit)")
    else:
        print("--- No trial data found in the database. ---")
```

```python
#Additional imports for experiments
#!pip install xgboost

import numpy as np
import pandas as pd
import time
import sqlite3
from joblib import Parallel, delayed
from IPython.display import display

import xgboost as xgb
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score

# Import your project files
from foldrm import Classifier
from utils import split_data, split_xy, get_scores, count_rules_in_model
from datasets import annaburroo
from algo import prune_rules

# --- Experiment Configuration ---
datasets_with_names = [(annaburroo, "Annaburroo")]

# Define the fit methods and strategies to test
fit_methods = ["FOLD-RM (fit)", "CON-FOLD (confidence_fit)"]
selection_strategies = ['greedy', 'round_robin', 'best_rule', 'info_gain']

metric_list = ['original', 'original_guarded', 'original_reweighted', 'original_t_removed', 'IG_over_P', 'IG_over_P2', 'LogOddsRatio',  'MathewsCC',
               'MathewsCC_mod', 'MathewsP', 'Gini_Impurity', 'Precision_Gini_Impurity']

NUM_TRIALS = 300 
DB_FILE = 'annaburroo_concise_exp80precent_training_pruning.db'

# Display options for the final DataFrame
pd.set_option('display.max_rows', 500)
pd.set_option('display.max_columns', 50)
pd.set_option('display.width', 1000)
```

```python
#Experiment with pruning
#Helper Functions for Experiments
def run_one_trial(dataset_name, strategy, fit_method, metric, data, model_template, num_classes, db_file):
    start_time = time.time()
    
    # 50/50 Train/Test split
    data_train, data_test = split_data(data, ratio=0.8) 
    X_train, Y_train = split_xy(data_train)
    X_test, Y_test = split_xy(data_test)
    
    rule_count = 0
    score = 0
    
    if fit_method == "XGBoost":
        # Ensure X features are purely numeric (stripping out string IDs like FireID)
        X_train_num = [[float(x) for x in row if isinstance(x, (int, float)) or (isinstance(x, str) and x.replace('.','',1).replace('-','',1).isdigit())] for row in X_train]
        X_test_num = [[float(x) for x in row if isinstance(x, (int, float)) or (isinstance(x, str) and x.replace('.','',1).replace('-','',1).isdigit())] for row in X_test]
        
        # Encode string labels to integers for XGBoost
        le = LabelEncoder()
        # Extract target strings whether they are tuples or bare strings
        y_train_flat = [y[0] if isinstance(y, tuple) else y for y in Y_train]
        y_test_flat = [y[0] if isinstance(y, tuple) else y for y in Y_test]
        
        le.fit(y_train_flat + y_test_flat)
        y_tr_enc = le.transform(y_train_flat)
        y_te_enc = le.transform(y_test_flat)
        
        # Train and Predict with default parameters
        clf = xgb.XGBClassifier(eval_metric='mlogloss')
        clf.fit(np.array(X_train_num), y_tr_enc)
        y_pred_enc = clf.predict(np.array(X_test_num))
        
        score = accuracy_score(y_te_enc, y_pred_enc)
        rule_count = 0 # XGBoost uses trees, so we default to 0 for the rule output
        
    else:
        # FOLD-RM and CON-FOLD Logic
        model = Classifier(attrs=model_template.attrs, numeric=model_template.numeric, label=model_template.label)
        
        if fit_method == "CON-FOLD (confidence_fit)":
            model.confidence_fit(data_train, metric='MathewsCC_mod', num_classes=num_classes, selection_strategy='best_rule', improvement_threshold=0.02)
        else:
            model.fit(data_train, metric=metric, num_classes=num_classes, selection_strategy=strategy)

        model.rules = prune_rules(model.rules, confidence=0.55)
        if hasattr(model, 'crs') and model.crs:
            model.crs = prune_rules(model.crs, confidence=0.55)
        
        Ystar_test_tuples = model.predict(X_test)
        Ystar_test = [y[0] if y[0] is not None else 'Unknown' for y in Ystar_test_tuples]
        
        score = get_scores(Ystar_test, data_test)
        rule_count = count_rules_in_model(model)
        
    elapsed_time = time.time() - start_time
    
    # Save to database
    conn = sqlite3.connect(db_file)
    cursor = conn.cursor()
    cursor.execute('PRAGMA journal_mode=WAL;')
    cursor.execute('''
        INSERT INTO trials (dataset, strategy, fit_method, metric, accuracy, time, num_rules)
        VALUES (?, ?, ?, ?, ?, ?, ?)
    ''', (dataset_name, strategy, fit_method, metric, score, elapsed_time, rule_count))
    conn.commit()
    conn.close()

def run_experiments_for_combination(dataset_info, strategy, fit_method, metric, remaining_trials, db_file):
    dataset_func, name = dataset_info
    print(f"--- Starting: Method='{fit_method}', Metric='{metric}', Strategy='{strategy}', Remaining={remaining_trials} ---")
    
    model_template, data = dataset_func()
    num_classes = len(pd.unique(pd.DataFrame(data).iloc[:, -1]))
    
    if remaining_trials > 0:
        Parallel(n_jobs=-1, verbose=0)(
            delayed(run_one_trial)(name, strategy, fit_method, metric, data, model_template, num_classes, db_file)
            for _ in range(remaining_trials)
        )
```

```python
#Display Results
def format_with_ste(mean_val, std_val, count):
    """Formats values as $mean \pm STE$."""
    ste = std_val / np.sqrt(count)
    return f"${mean_val:.4f} \pm {ste:.4f}$"

def process_and_display_concise(df, xgb_df, fit_method_name):
    if df.empty:
        return

    # Aggregate Data
    agg = df.groupby(['metric', 'strategy']).agg({
        'accuracy': ['mean', 'std', 'count'],
        'num_rules': ['mean', 'std', 'count'],
        'time': ['mean', 'std', 'count']
    })

    # Format values with STE
    for col in ['accuracy', 'num_rules', 'time']:
        agg[(col, 'formatted')] = agg.apply(lambda row: format_with_ste(row[(col, 'mean')], row[(col, 'std')], row[(col, 'count')]), axis=1)

    # Pivot Tables
    acc_pivot = agg['accuracy']['formatted'].unstack(level='strategy')
    rules_pivot = agg['num_rules']['formatted'].unstack(level='strategy')
    time_pivot = agg['time']['formatted'].unstack(level='strategy')
    
    # Order the columns safely
    ordered_cols = [s for s in selection_strategies if s in acc_pivot.columns]
    acc_pivot = acc_pivot[ordered_cols]
    rules_pivot = rules_pivot[ordered_cols]
    time_pivot = time_pivot[ordered_cols]

    # Combine into wide format
    final_table = pd.concat([acc_pivot, rules_pivot, time_pivot], axis=1, keys=['Accuracy', 'Number of Rules', 'Time'])

    # Process and append XGBoost Control Row
    if not xgb_df.empty:
        xgb_agg = xgb_df.agg({
            'accuracy': ['mean', 'std', 'count'],
            'num_rules': ['mean', 'std', 'count'],
            'time': ['mean', 'std', 'count']
        })
        
        xgb_row = {}
        for top_col in ['Accuracy', 'Number of Rules', 'Time']:
            db_col = top_col.lower().replace('number of rules', 'num_rules')
            
            # Use standard formatting for rules vs accuracy/time
            if db_col == 'num_rules':
                val_str = "-" # XGBoost has no rules
            else:
                val_str = format_with_ste(xgb_agg.loc['mean', db_col], xgb_agg.loc['std', db_col], xgb_agg.loc['count', db_col])
                
            for strat in ordered_cols:
                xgb_row[(top_col, strat)] = val_str # Duplicate across row for easy visual comparison
                
        xgb_series = pd.Series(xgb_row, name="XGBoost (Control)")
        final_table = pd.concat([final_table, xgb_series.to_frame().T])

    print(f"\n{'='*60}")
    print(f"--- STRATEGY EXPERIMENT SUMMARY: {fit_method_name} ---")
    
    # Display table (Jupyter will automatically render the LaTeX \pm signs)
    display(final_table)
    print(f"{'='*60}")

if __name__ == "__main__":
    conn = sqlite3.connect(DB_FILE)
    try:
        all_trials_df = pd.read_sql_query("SELECT * FROM trials", conn)
    except pd.io.sql.DatabaseError:
        all_trials_df = pd.DataFrame()
    conn.close()

    if not all_trials_df.empty:
        # Separate out the XGBoost data
        xgb_df = all_trials_df[all_trials_df['fit_method'] == 'XGBoost']
        
        # Process FOLD-RM
        fold_rm_df = all_trials_df[all_trials_df['fit_method'] == 'FOLD-RM (fit)']
        process_and_display_concise(fold_rm_df, xgb_df, "FOLD-RM (fit)")
        
        # Process CON-FOLD
        con_fold_df = all_trials_df[all_trials_df['fit_method'] == 'CON-FOLD (confidence_fit)']
        process_and_display_concise(con_fold_df, xgb_df, "CON-FOLD (confidence_fit)")
    else:
        print("--- No trial data found in the database. ---")
```

```python
import pandas as pd
from datasets import annaburroo
from utils import split_data
from foldrm import Classifier
from algo import prune_rules

# 1. Setup and Load Data
model_template, data = annaburroo()
num_classes = len(pd.unique(pd.DataFrame(data).iloc[:, -1]))
features = model_template.attrs

# 2. Split Data (80% Train, 20% Test)
data_train, data_test = split_data(data, ratio=0.8)

# 3. Initialize and Train the Model
model = Classifier(attrs=model_template.attrs, numeric=model_template.numeric, label=model_template.label)

print("Training CON-FOLD model with 'MathewsCC_mod' and 'best_rule'...")
model.confidence_fit(data_train, metric='MathewsCC_mod', num_classes=num_classes, selection_strategy='best_rule', improvement_threshold=0.02)

model.rules = prune_rules(model.rules, confidence=0.70)


# Finalize the rule set
if hasattr(model, 'asp'):
    model.asp()

# 4. Print the Generated Rules
print("\n--- Learned Rules (MathewsCC_mod + best_rule) ---")

# Pull from 'crs' (confidence rules) if it exists, otherwise fallback to standard 'rules'
rules_to_print = model.crs if hasattr(model, 'crs') and model.crs else model.rules

if rules_to_print:
    rule_counter = 1
    for rule in rules_to_print:
        # Check if confidence is the last element
        if len(rule) > 3 and isinstance(rule[-1], (float, int)):
            conf = rule[-1]
            conf_str = f" (Confidence: {conf:.4f})"
        else:
            conf = 0.0
            conf_str = ""

        # SKIP internal exception rules generated by FOLD (confidence 0)
        if conf <= 0.0:
            continue

        # Format the body (conditions)
        formatted_conditions = []
        for cond in rule[1]:
            feat_idx, operator, value = cond
            
            # Map numeric index to column name
            feat_name = features[feat_idx] if isinstance(feat_idx, int) else feat_idx
            formatted_conditions.append(f"{feat_name} {operator} {value}")
        
        body_str = " AND ".join(formatted_conditions) if formatted_conditions else "True"
        
        if isinstance(rule[0], tuple) and len(rule[0]) >= 3:
            class_name = rule[0][2]
        else:
            class_name = str(rule[0]) 
            
        head_str = f"Class = '{class_name}'"
            
        print(f"Rule {rule_counter}: IF {body_str} THEN {head_str}{conf_str}")
        rule_counter += 1
else:
    print("No rules were generated during this trial.")
```
